```python
import pandas as pd

# 1. Đọc dữ liệu thô (Raw Data)
movies_raw = pd.read_csv('./dataset/ml-20m/movies.csv')
ratings_raw = pd.read_csv('./dataset/ml-20m/ratings.csv')

# 2. Làm sạch dữ liệu (Clean Data)
movies_clean = movies_raw.dropna().drop_duplicates()
ratings_clean = ratings_raw.dropna().drop_duplicates()
# Lọc điểm đánh giá hợp lệ từ 0.5 đến 5.0
ratings_clean = ratings_clean[(ratings_clean['rating'] >= 0.5) & (ratings_clean['rating'] <= 5.0)]

# 3. Tổng hợp số liệu
summary_data = {
    'Tệp dữ liệu': ['movies.csv', 'ratings.csv'],
    'Trước khi làm sạch (Dòng)': [f"{len(movies_raw):,}", f"{len(ratings_raw):,}"],
    'Sau khi làm sạch (Dòng)': [f"{len(movies_clean):,}", f"{len(ratings_clean):,}"],
    'Số dòng đã loại bỏ': [f"{len(movies_raw) - len(movies_clean):,}", f"{len(ratings_raw) - len(ratings_clean):,}"]
}

# 4. Xuất thông tin ra bảng dạng Pandas DataFrame
df_summary = pd.DataFrame(summary_data)

print("=== THỐNG KÊ SỐ LƯỢNG DÒNG DỮ LIỆU ===")
display(df_summary)
```

```python
# Tạo biểu đồ đánh giá RMSE của mô hình ALS

import matplotlib.pyplot as plt

# Dữ liệu mẫu đánh giá RMSE (Thay bằng số liệu thực tế bạn train được)
ranks = [5, 10, 15, 20]
rmse_values = [0.885, 0.832, 0.815, 0.812]

plt.figure(figsize=(8, 4.5))
# Đã thêm dấu nháy đơn ' ' bao quanh mã màu #3498db
plt.plot(ranks, rmse_values, marker='o', color='#3498db', linewidth=2.5, markersize=8)

plt.title('Đánh giá chỉ số RMSE của mô hình ALS theo Rank', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Tham số Rank (Số chiều thuộc tính ẩn)', fontsize=11)
plt.ylabel('Chỉ số RMSE (Càng thấp càng tốt)', fontsize=11)
plt.xticks(ranks)
plt.grid(True, linestyle='--', alpha=0.6)

# Hiển thị giá trị cụ thể trên từng điểm
for i, txt in enumerate(rmse_values):
    plt.annotate(f"{txt}", (ranks[i], rmse_values[i]), textcoords="offset points", xytext=(0,10), ha='center')

plt.tight_layout()
plt.show()
```
